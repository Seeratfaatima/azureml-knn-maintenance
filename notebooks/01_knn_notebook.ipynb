### Cell 2: imports and data

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split, StratifiedKFold, GridSearchCV
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score,
                             roc_auc_score, classification_report, ConfusionMatrixDisplay)

df = pd.read_csv("../data/ai4i_clean.csv")
TARGET = "machine_failure"
NUM_COLS = ["air_temp_k", "process_temp_k", "rpm", "torque_nm", "tool_wear_min"]
CAT_COLS = ["type"]
df[NUM_COLS] = df[NUM_COLS].astype(float)
df.head()
```

### Cell 3: how rare are failures?

```python
fig, axes = plt.subplots(1, 2, figsize=(11, 4))
df[TARGET].value_counts().plot(kind="bar", ax=axes[0], color=["steelblue", "tomato"])
axes[0].set_title("Count of machines (0 = OK, 1 = failed)")
df.groupby("type")[TARGET].mean().mul(100).plot(kind="bar", ax=axes[1], color="gray")
axes[1].set_title("Failure rate (%) by product type")
plt.tight_layout()
plt.show()
```

### Cell 4: which sensors differ for failed machines?

```python
fig, axes = plt.subplots(1, 5, figsize=(20, 4))
for ax, col in zip(axes, NUM_COLS):
    sns.boxplot(data=df, x=TARGET, y=col, ax=ax)
    ax.set_title(col)
plt.tight_layout()
plt.show()
```

### Cell 6: Split into training and test sets

```python
X = df[NUM_COLS + CAT_COLS]
y = df[TARGET]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, stratify=y, random_state=42
)
print("Train:", X_train.shape, "failure rate:", round(y_train.mean(), 4))
print("Test: ", X_test.shape, "failure rate:", round(y_test.mean(), 4))
```

### Cell 7 (Proof that Scaling Matters):

```python
def make_pipeline(scale=True, k=5):
    num_step = StandardScaler() if scale else "passthrough"
    prep = ColumnTransformer([
        ("num", num_step, NUM_COLS),
        ("cat", OneHotEncoder(handle_unknown="ignore"), CAT_COLS),
    ])
    return Pipeline([("prep", prep), ("knn", KNeighborsClassifier(n_neighbors=k))])

for scale in [False, True]:
    model = make_pipeline(scale=scale).fit(X_train, y_train)
    pred = model.predict(X_test)
    print(f"Scaling={str(scale):5} accuracy={accuracy_score(y_test, pred):.3f} "
          f"recall={recall_score(y_test, pred):.3f} F1={f1_score(y_test, pred):.3f}")
```

### Why Accuracy Remains Similar While Recall Changes Significantly

1. **Extreme Class Imbalance Masks Minority Class Performance**:
   In the AI4I dataset, machine failures account for only ~3.4% of total observations. A model predicting the majority class ("no failure") for virtually every sample naturally scores around 97% accuracy. Because accuracy treats all errors equally across classes, correct predictions on the dominant negative class swamp the metric.

2. **Distance Distortion in Unscaled KNN**:
   K-Nearest Neighbors relies entirely on Euclidean distance calculations. Without feature scaling (`Scaling=False`), features with large absolute numerical values and variances (such as `rpm` at ~1500 or `air_temp_k` at ~300) dominate the distance metric. Subtle variations in smaller-scale features (such as `tool_wear_min` or `torque_nm`), which are strong indicators of mechanical failure, are effectively ignored. As a result, the unscaled model rarely identifies true positive failure instances, driving recall down to **0.206**.

3. **Feature Scaling Enables Failure Detection**:
   When features are normalized (`Scaling=True`), each feature contributes proportionately to distance calculations. This allows KNN to detect the local feature combinations characteristic of breakdowns, increasing recall from **0.206 to 0.294** (a ~43% relative improvement) and F1-score from **0.318 to 0.435**, while baseline accuracy appears virtually unchanged (0.970 vs 0.974).


### Cell 8 & 9 (Hyperparameter Tuning via GridSearchCV):

```python
param_grid = {
    "knn__n_neighbors": [1, 3, 5, 7, 9, 11, 15, 21],
    "knn__weights": ["uniform", "distance"],
    "knn__p": [1, 2],
}
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
grid = GridSearchCV(make_pipeline(scale=True), param_grid, scoring="f1", cv=cv, n_jobs=-1)
grid.fit(X_train, y_train)
print("Best parameters:", grid.best_params_)
print("Best cross-validated F1:", round(grid.best_score_, 3))
```

```python
res = pd.DataFrame(grid.cv_results_)
res = res[res["param_knn__p"] == grid.best_params_["knn__p"]]
for w in ["uniform", "distance"]:
    sub = res[res["param_knn__weights"] == w]
    plt.plot(sub["param_knn__n_neighbors"].astype(int), sub["mean_test_score"], marker="o", label=w)
plt.xlabel("K (number of neighbours)")
plt.ylabel("Mean cross-validated F1")
plt.title("Choosing K")
plt.legend()
plt.grid(True)
plt.show()
```

### Cell 10 & 11 (Evaluation & Decision Thresholding):

```python
best_model = grid.best_estimator_
y_pred = best_model.predict(X_test)
y_prob = best_model.predict_proba(X_test)[:, 1]

metrics = {
    "accuracy": accuracy_score(y_test, y_pred),
    "precision": precision_score(y_test, y_pred),
    "recall": recall_score(y_test, y_pred),
    "f1": f1_score(y_test, y_pred),
    "auc": roc_auc_score(y_test, y_prob),
}
print({k: round(v, 3) for k, v in metrics.items()})
print(classification_report(y_test, y_pred, target_names=["OK", "Failure"]))

fig, ax = plt.subplots(figsize=(4, 4))
ConfusionMatrixDisplay.from_predictions(y_test, y_pred, display_labels=["OK", "Failure"], cmap="Blues", ax=ax)
ax.set_title("KNN confusion matrix (test set)")
fig.savefig("confusion_matrix.png", bbox_inches="tight")
plt.show()
```

```python
for t in [0.5, 0.4, 0.3, 0.2]:
    p = (y_prob >= t).astype(int)
    print(f"threshold={t}: recall={recall_score(y_test, p):.3f} precision={precision_score(y_test, p, zero_division=0):.3f}")
```

### Threshold Recommendation for the Maintenance Manager

#### Recommendation:
I recommend maintaining an operational decision threshold of **0.5** (or selecting the highest threshold within the identical-metric range: **0.5**).

---

### Justification and Analysis:

1. **Empirical Performance Invariance ($0.2$ to $0.5$):**
   * Evaluating thresholds across `[0.5, 0.4, 0.3, 0.2]` yields identical outcomes: **Recall = 0.353** and **Precision = 0.453**.
   * In a KNN classifier (such as $k=5$), predicted probabilities are discrete increments corresponding to the proportion of nearest neighbors (e.g., $0.0, 0.2, 0.4, 0.6, \dots$). No test instances fall into the probability window between $0.2$ and $0.5$. Lowering the threshold down to $0.2$ produces zero gain in catching additional true failures while leaving precision unchanged.

2. **Operational Cost-Benefit Balance:**
   * **Cost of False Negatives (Missed Failures):** Unplanned machine downtime, secondary component damage, and production halts carry high financial and safety penalties. Higher recall is typically preferred in predictive maintenance to avoid catastrophic line stoppages.
   * **Cost of False Positives (False Alarms):** Unnecessary technician dispatch and diagnostic inspections consume labor hours and disrupt normal operations.
   * At a threshold of **0.5**, the model identifies true failures with a precision of **45.3%** without taking on unnecessary false positives from even lower thresholds (e.g., $<0.2$).

3. **Strategic Next Steps for Management:**
   * Because KNN probability granularity is coarse, adjusting the decision threshold alone cannot improve recall beyond **0.353** without drastically diluting precision.
   * To achieve a higher recall target (e.g., $>80\%$) suitable for high-consequence machinery, recommend transitioning to algorithms better suited for severe class imbalances, such as **Cost-Sensitive Learning**, **XGBoost/LightGBM with class weights**, or **AutoML**, rather than solely relying on KNN threshold shifts.
